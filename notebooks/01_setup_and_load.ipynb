```python
import re
from collections import Counter 

descriptions = df["description"].dropna().astype(str)
text = " ".join(descriptions).lower()
words = re.findall(r"\b[a-zA-Z]+\b",text)

stopwords = set([
    "the","and","of","to","in","a","an","for","on","with","is","as","at","by","from","this","that","it","its","his",
    "her","their","who","what","when","where","into","after","before","about"
])

words_clean = [w for w in words if w not in stopwords and len(w) > 2]
word_freq = Counter(words_clean)
top_words = word_freq.most_common(20)

top_words
```

```python
import pandas as pd
import matplotlib.pyplot as plt

top_df = pd.DataFrame(top_words , columns= ["word","count"])

plt.figure(figsize=(10,5))
plt.bar(top_df["word"],top_df["count"])
plt.xticks(rotation=45, ha="right")
plt.title("Netflix İçerik Açıklamalarında En Sık Geçen 20 Kelime")
plt.xlabel("Word")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
def get_top_words(df,type_name, top_n=15):
    desc = df[df["type"] == type_name]["description"].dropna().astype(str)
    text = " ".join(desc).lower()
    words = re.findall(r"\b[a-zA-Z]+\b",text)
    words = [w for w in words if w not in stopwords and len(w)  > 2]
    return Counter(words).most_common(top_n)

top_movie = get_top_words(df,"Movie")
top_tv = get_top_words(df,"TV Show")

top_movie, top_tv
```

```python
movie_df = pd.DataFrame(top_movie, columns=["word","count"])
tv_df = pd.DataFrame(top_tv,columns=["word","count"])

plt.figure(figsize=(10,5))
plt.bar(movie_df["word"],movie_df["count"])
plt.xticks(rotation=45, ha="right")
plt.title("Film İçerik Açıklamalarında En Sık Geçen Kelimeler")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10,5))
plt.bar(tv_df["word"],tv_df["count"])
plt.xticks(rotation=45, ha="right")
plt.title("Dizi İçerik Açıklamalarında En Sık Geçen Kelimeler")
plt.tight_layout()
plt.show()
```
